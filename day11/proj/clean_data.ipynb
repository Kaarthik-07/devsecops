```python
import pandas as pd

#inspect the dataset
df = pd.read_csv("sales_dirty.csv")
print("------------------------------data types and total rows of non null value data------------------------------------------")
df.info()
print("------------------------------gives all the staustical data of mean count std max everyhitng that needs for computation or analaysis ------------------------------------------")
print(df.describe())
print("------------------------------gives all the columns that are present in the csv or dataset ------------------------------------------")
print(df.columns)
print("------------------------------total count of null values on each column ------------------------------------------")
print(df.isnull().sum())
```

```python
#identify and remove duplicates

print(df.duplicated().sum())
print(df[df.duplicated()])

df = df.drop_duplicates()
df
```

```python
# Standardize product and city names
df["Product"] = df["Product"].str.strip().str.title()
df["City"] = df["City"].str.strip().str.title()
print("------------------------------Standardized value for products no inconsistent data for product column------------------------------------------")
print(df["product"])
print("------------------------------Standardized value for products no inconsistent data for city column------------------------------------------")

print(df["City"])
```

```python
# convert data into a proper data type
df["OrderDate"] = pd.to_datetime(df["OrderDate"],errors="coerce") 
print(df[["OrderDate"]])
print(df["OrderDate"].dtype)
```

```python
# convert numeric columns
df["Quantity"] = pd.to_numeric(df["Quantity"], errors = "coerce")
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors = "coerce")
print(df["Quantity"])
```

```python
# handle missing values
df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)

df["UnitPrice"] = df["UnitPrice"].fillna(
    df["UnitPrice"].median()
)

df["City"] = df["City"].fillna("Unknown")

print(df["Quantity"])
print(df["City"])
```

```python
# price category for each unit price to know where outlier might have occured
def priceCategory(price):
    if pd.isna(price):
        return "Unknown"
    elif price < 1000:
        return "Low"
    elif price < 10000:
        return "Medium"
    else:
        return "High"

df["PriceCategory"] = df["UnitPrice"].apply(priceCategory)

print(df[["UnitPrice", "PriceCategory"]])
```

```python
# calcualte revewnue 
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print(df[["Product", "Quantity", "UnitPrice", "Revenue"]])
```
