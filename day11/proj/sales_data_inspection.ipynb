```python
import pandas as pd

df = pd.read_csv("sales_dirty.csv")

# idenitfying missing values in every columns
print(df.isnull().sum())
```

```python
# identify duplicate orders

print("Duplicated Orders", df.duplicated().sum())
print(df[df.duplicated()])
```

```python
# identify duplicated order ids
print("Duplicated orderid : ", df["OrderID"].duplicated().sum())
print(df[df["OrderID"].duplicated(keep=False)])
```

```python
# find five highest priced products
print(df.nlargest(5,"UnitPrice")[["Product","UnitPrice"]])
```

```python
# find the highest available unit price
print("Highest UnitPrice:", df["UnitPrice"].max())
```
