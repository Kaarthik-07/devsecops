```python
import pandas as pd

df = pd.read_csv("sales_dirty.csv")
df
```

```python
# Inspect the dataset
print(df.head())
print(df.columns)
print(df.isnull().sum())
```

```python
# Convert quantity and unit price into numeric values
df["Quantity"] = pd.to_numeric(
    df["Quantity"], errors="coerce"
)

df["UnitPrice"] = pd.to_numeric(
    df["UnitPrice"], errors="coerce"
)
print(df["Quantity"])
print(df["UnitPrice"])
```

```python
#Revenue for each record
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print(df[["City", "Quantity", "UnitPrice", "Revenue"]])
```

```python
# gropu total revenue by city
city_revenue = df.groupby("City")["Revenue"].sum()

print(city_revenue.sort_values(ascending=False))
```

```python
# identify highest revenue by city
print("Highest Revenue City:", city_revenue.idxmax())
print("Highest City Revenue:", city_revenue.max())
```

```python
# apply to a new csv
df.to_csv("sales_revenue_analysis.csv", index=False)
```
